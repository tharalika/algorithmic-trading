```python
import yfinance as yf
import plotly.graph_objects as go
import pandas as pd

def run_algo_strategy():
    print("Fetching market data and calculating moving averages...")
    # Fetch 1 year of data for the BSE Sensex (Indian Market Index)
    ticker = "^BSESN"
    df = yf.download(ticker, period="1y", interval="1d", progress=False)

    # Flatten multi-level columns if they exist (yfinance quirk)
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.get_level_values(0)

    # Calculate Moving Averages
    df['SMA_20'] = df['Close'].rolling(window=20).mean()
    df['SMA_50'] = df['Close'].rolling(window=50).mean()

    # Create the interactive Candlestick chart
    fig = go.Figure()

    fig.add_trace(go.Candlestick(x=df.index, open=df['Open'], high=df['High'],
                                 low=df['Low'], close=df['Close'], name='Market Price'))

    fig.add_trace(go.Scatter(x=df.index, y=df['SMA_20'], line=dict(color='blue', width=1.5), name='20-Day SMA'))
    fig.add_trace(go.Scatter(x=df.index, y=df['SMA_50'], line=dict(color='orange', width=1.5), name='50-Day SMA'))

    fig.update_layout(title="Algorithmic Trading: 20-Day vs 50-Day SMA Crossover",
                      yaxis_title="Index Value", template="plotly_dark", xaxis_rangeslider_visible=False)
    return fig

# Run and display
algo_chart = run_algo_strategy()
algo_chart.show()
```
